```python
import itertools

def is_even(perm):
    """Returns True if the permutation is even."""
    inv = 0
    for i in range(len(perm)):
        for j in range(i + 1, len(perm)):
            if perm[i] > perm[j]:
                inv += 1
    return inv % 2 == 0

def to_compressed_cycles(perm):
    """Converts a 0-indexed permutation to compressed cycle notation."""
    p = [x + 1 for x in perm]
    visited = [False] * len(p)
    cycles = []
    for i in range(len(p)):
        if not visited[i]:
            cycle = []
            curr = i
            while not visited[curr]:
                visited[curr] = True
                cycle.append(str(p[curr]))
                curr = p[curr] - 1
            if len(cycle) > 1:
                cycles.append(f"[{''.join(cycle)}]")
    return "".join(cycles) if cycles else "(1)"

def perm_mult(p1, p2):
    """Multiplies two permutations (composition p1 * p2)."""
    return tuple(p1[p2[i]] for i in range(len(p1)))

def perm_inv(p):
    """Computes the inverse of a permutation."""
    pi = [0] * len(p)
    for i, x in enumerate(p):
        pi[x] = i
    return tuple(pi)

def get_subgroup(gens, n):
    """Generates a subgroup given a set of generator permutations."""
    sg = set(gens)
    sg.add(tuple(range(n)))
    while True:
        new_elements = set()
        for x in sg:
            for y in sg:
                m = perm_mult(x, y)
                if m not in sg:
                    new_elements.add(m)
        if not new_elements:
            break
        sg.update(new_elements)
    return sg

def find_all_sylows(base_subgroup, G):
    """Finds all unique Sylow subgroups by conjugating a base subgroup across G."""
    all_sylows = set()
    for g in G:
        g_inv = perm_inv(g)
        conjugated = tuple(sorted(perm_mult(g, perm_mult(x, g_inv)) for x in base_subgroup))
        all_sylows.add(conjugated)
    return sorted(list(all_sylows))

# 1. Setup A_5 and establish tracking for its 5-Sylow subgroups to define the f mapping
A5 = [p for p in itertools.permutations(range(5)) if is_even(p)]
gen_5_a5 = (1, 2, 3, 4, 0)
sylow_5_a5 = find_all_sylows(get_subgroup([gen_5_a5], 5), A5)

# 2. Map f(x) : A_5 -> A_6 using the index of the 5-Sylows
f_map = {}
for x in A5:
    x_inv = perm_inv(x)
    perm_of_S = []
    for s in sylow_5_a5:
        conj_s = tuple(sorted(perm_mult(x, perm_mult(y, x_inv)) for y in s))
        perm_of_S.append(sylow_5_a5.index(conj_s))
    f_map[x] = tuple(perm_of_S)

# 3. Setup A_6 and its Sylow subgroups
A6 = [p for p in itertools.permutations(range(6)) if is_even(p)]

sylow_2_a6 = find_all_sylows(get_subgroup([(1, 2, 3, 0, 5, 4), (2, 1, 0, 3, 5, 4)], 6), A6)
sylow_3_a6 = find_all_sylows(get_subgroup([(1, 2, 0, 3, 4, 5), (0, 1, 2, 4, 5, 3)], 6), A6)
sylow_5_a6 = find_all_sylows(get_subgroup([(1, 2, 3, 4, 0, 5)], 6), A6)

# 4. Helper function to process and print results with specific matching elements
def print_processed_sylows(sylow_list, current_idx):
    for sylow in sylow_list:
        formatted = [to_compressed_cycles(p) for p in sylow]
        matching_elements = []
        
        # Check every non-identity element x from A5
        for x in A5:
            if x != (0, 1, 2, 3, 4) and f_map[x] in sylow:
                matching_elements.append(to_compressed_cycles(x))
                
        suffix = ""
        if matching_elements:
            # Sort for readability and join elements with commas
            matching_elements.sort()
            suffix = f" -> {', '.join(matching_elements)}"
            
        print(f"{current_idx:02d}: [{', '.join(formatted)}]{suffix}")
        current_idx += 1
    return current_idx

# --- Run Outputs ---
global_idx = 1
print(f"=== 2-SYLOW SUBGROUPS OF A_6 (Total: {len(sylow_2_a6)}) ===")
global_idx = print_processed_sylows(sylow_2_a6, global_idx)

print(f"\n=== 3-SYLOW SUBGROUPS OF A_6 (Total: {len(sylow_3_a6)}) ===")
global_idx = print_processed_sylows(sylow_3_a6, global_idx)

print(f"\n=== 5-SYLOW SUBGROUPS OF A_6 (Total: {len(sylow_5_a6)}) ===")
global_idx = print_processed_sylows(sylow_5_a6, global_idx)
```
